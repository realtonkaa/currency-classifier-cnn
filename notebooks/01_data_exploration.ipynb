# Data Exploration — Currency Classifier CNN

In this notebook, I explore the banknote dataset to understand:
- How many images per currency/denomination
- Image quality and dimensions
- Class balance
- Sample visualizations

```python
import os
from pathlib import Path
from collections import Counter
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

DATA_DIR = Path("../data")
print(f"Looking for data in: {DATA_DIR.resolve()}")
```

```python
# Count images per class
class_counts = {}
if DATA_DIR.exists():
    for class_dir in sorted(DATA_DIR.iterdir()):
        if class_dir.is_dir() and not class_dir.name.startswith("."):
            count = len(list(class_dir.glob("*.jpg")) + list(class_dir.glob("*.png")))
            class_counts[class_dir.name] = count

if class_counts:
    print(f"Total classes: {len(class_counts)}")
    print(f"Total images: {sum(class_counts.values())}")
    print(f"\nImages per class:")
    for cls, count in sorted(class_counts.items()):
        print(f"  {cls}: {count}")
else:
    print("No data found yet. Download the dataset first.")
    print("See README.md for instructions.")
```

## Class Distribution

A balanced dataset is important for training. Let's visualize the distribution.

```python
if class_counts:
    fig, ax = plt.subplots(figsize=(12, 6))
    currencies = list(class_counts.keys())
    counts = list(class_counts.values())

    ax.barh(currencies, counts, color='steelblue')
    ax.set_xlabel('Number of Images')
    ax.set_title('Images per Currency Class')
    ax.grid(axis='x', alpha=0.3)
    plt.tight_layout()
    plt.show()
else:
    print("No data to visualize yet.")
```

## Sample Images

Let's look at some example banknotes from the dataset.

```python
if class_counts:
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    axes = axes.flatten()

    for i, (cls_name, _) in enumerate(list(class_counts.items())[:8]):
        cls_dir = DATA_DIR / cls_name
        images = list(cls_dir.glob("*.jpg")) + list(cls_dir.glob("*.png"))
        if images:
            img = Image.open(images[0])
            axes[i].imshow(img)
            axes[i].set_title(cls_name, fontsize=10)
            axes[i].axis("off")

    plt.suptitle("Sample Banknotes from Each Class", fontsize=14)
    plt.tight_layout()
    plt.show()
else:
    print("No data to visualize yet.")
```
