# Training Experiments — Currency Classifier CNN

Documenting my experiments with different model configurations:
1. Basic CNN from scratch (baseline)
2. MobileNetV2 with frozen backbone
3. MobileNetV2 with full fine-tuning

```python
import matplotlib.pyplot as plt
import numpy as np

# Results from my training runs
experiments = {
    "Basic CNN (scratch)": {"val_acc": 67, "epochs": 30, "time_min": 45},
    "MobileNetV2 (frozen)": {"val_acc": 89, "epochs": 15, "time_min": 20},
    "MobileNetV2 (fine-tuned)": {"val_acc": 94, "epochs": 25, "time_min": 35},
}

print("Experiment Results:")
print("-" * 55)
print(f"{'Model':<30} {'Val Acc':<10} {'Epochs':<10} {'Time'}")
print("-" * 55)
for name, results in experiments.items():
    print(f"{name:<30} {results['val_acc']}%{'':6} {results['epochs']:<10} {results['time_min']} min")
```

```python
# Visualize comparison
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

names = list(experiments.keys())
accs = [e["val_acc"] for e in experiments.values()]
times = [e["time_min"] for e in experiments.values()]

ax1.barh(names, accs, color=['#cc4444', '#4488cc', '#44cc88'])
ax1.set_xlabel("Validation Accuracy (%)")
ax1.set_title("Model Comparison — Accuracy")
ax1.set_xlim(0, 100)
ax1.grid(axis='x', alpha=0.3)

ax2.barh(names, times, color=['#cc4444', '#4488cc', '#44cc88'])
ax2.set_xlabel("Training Time (minutes)")
ax2.set_title("Model Comparison — Speed")
ax2.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()
```

## Key Findings

1. **Transfer learning is way better than training from scratch.** The pretrained MobileNetV2 started at ~70% accuracy on epoch 1, while the basic CNN took 20 epochs to reach that.

2. **Fine-tuning the full model gives the best results** but takes longer. The frozen backbone got to 89% much faster, which is a good tradeoff if you need quick results.

3. **MobileNetV2 is a good choice** because it's small enough to run on a phone (which matters for accessibility — the target users might have older phones).
